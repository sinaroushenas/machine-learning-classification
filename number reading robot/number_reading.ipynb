```python
from sklearn import linear_model
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
```

```python
from sklearn import datasets
```

```python
digits= datasets.load_digits()
```

```python
digits.data.shape
```

```python
digits.images.shape
```

```python
digits.target
```

```python
digits.target[3]
```

```python
plt.imshow(digits.images[3])
```

```python
x=digits.data
y= digits.target
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2)
```

```python
model= linear_model.LogisticRegression()
model.fit(x_train,y_train)
```

```python
out_robot=model.predict(x_test)
```

```python
out_robot
```

```python
y_test
```

```python
er= y_test - out_robot
print(er.shape)
print(er)
```

```python
sahih=0
qalat=0
for z in er:
    if z==0:
        sahih=sahih+1
    elif z!=0:
        qalat=qalat+1

print(sahih)
print(qalat)
```

```python
ac=model.score(x_test,y_test)
ac=ac*100
print("the robot acuracy is:",ac,"%")
```
